```python
import os
import pandas as pd
from pymongo import MongoClient
from dotenv import load_dotenv

# Carrega as variáveis de ambiente do arquivo .env (que está uma pasta para trás)
load_dotenv("../.env")

MONGO_USER = os.getenv("MONGO_INITDB_ROOT_USERNAME")
MONGO_PASSWORD = os.getenv("MONGO_INITDB_ROOT_PASSWORD")
MONGO_HOST = os.getenv("MONGO_HOST", "localhost")
MONGO_PORT = os.getenv("MONGO_PORT", "27017")
MONGO_DATABASE = os.getenv("MONGO_DATABASE", "panorama")
# Note que no script a variável padrão foi mudada para "reddit", altere se precisar para "reddit_teste"
MONGO_COLLECTION = os.getenv("MONGO_COLLECTION", "reddit") 

# Monta a URI e conecta
MONGO_URI = f"mongodb://{MONGO_USER}:{MONGO_PASSWORD}@{MONGO_HOST}:{MONGO_PORT}/"
client = MongoClient(MONGO_URI)
db = client[MONGO_DATABASE]
collection = db[MONGO_COLLECTION]

# Verifica o total de documentos gravados na coleção
total_docs = collection.count_documents({})
print(f"Total de documentos coletados em '{MONGO_COLLECTION}': {total_docs}")

# Puxa os 5 posts/comentários mais recentes e joga num DataFrame do Pandas para visualizar!
cursor = collection.find().sort("created_utc", -1).limit(5)
df = pd.DataFrame(list(cursor))

# Se você quiser remover a coluna '_id' padrão do Mongo da visualização:
if '_id' in df.columns:
    df = df.drop(columns=['_id'])

# Mostra a tabela bonita no notebook
df.head()
```
